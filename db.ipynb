# [dbo].[registerCommunity]

```sql
/****** Object:  StoredProcedure [dbo].[registerCommunity]    Script Date: 22-02-2026 02:23:49 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE proc [dbo].[registerCommunity]
@cname varchar(100),
@cpassword varchar(100),
@cimage varbinary(max),
@cemail varchar(100)
as 
begin
	if exists(select * from Community where @cemail=cemail)
	begin
		select -1;
	end
	else
	begin
		insert into Community values(@cname,@cpassword,@cimage,@cemail)
		select 1;
	end
end
GO
```
